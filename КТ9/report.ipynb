```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when
import os


spark = SparkSession.builder.getOrCreate()

dataset_path = "C:/Users/User/Desktop/jupyterhub-on-hadoop/datasets"
files = [f for f in os.listdir(dataset_path)]
dfs = [spark.read.csv(dataset_path + '/' + file, header=True, sep=';') for file in files]
dfs = [df.dropDuplicates(["unified_id"]) for df in dfs if "unified_id" in df.columns]
dfs = [df.na.drop("any") for df in dfs]
```

```python
result = dfs[0]
for df in dfs[1:]: result = df.join(df, "unified_id", 'inner')
result.show(10, truncate=False)
```
